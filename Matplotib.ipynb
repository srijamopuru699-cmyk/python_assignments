```python
1. Import Required Libraries

In Python, we usually import matplotlib.pyplot to draw charts and pandas to work with data tables. Notes pyplot is the plotting interface used for most Matplotlib charts. pandas helps store and inspect tabular data. In Jupyter Notebook, plots appear inline when the notebook is configured correctly.
```

```python
import pandas as pd
import matplotlib.pyplot as plt
print('matplotlib is ready!')
```

```python
2. Load and Inspect Sample Data

Let's create a small dataset in pandas and inspect it before plotting. Notes head() shows the first rows of the dataset. shape tells how many rows and columns exist. describe() gives summary statistics like mean, minimum, and maximum. Before plotting, it is important to understand the structure and meaning of the data.


```

```python
sales_data = pd.DataFrame({
    'month' : ['jan','feb','mar','apr','may','jun'],
    'sales' : [120,150,90,180,260,210]
})
print(sales_data)
print('\nShape:', sales_data.shape)
print('\nColumns:', sales_data.columns.tolist())
print('\nHead:')
print(sales_data.head())
print('\nDescription:')
print(sales_data.describe())


```

 Create a Simple Line Plot

A line chart is useful when showing trends over time or ordered values. Notes plt.plot() draws a line between points. marker='o' adds circle markers at each value. The line shows the trend in the data. Line plots are best when you want to observe change over time.


```python
plt.figure(figsize=(8,5))
plt.plot(sales_data['month'],sales_data['sales'], color='green', marker='o',linewidth=2)
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.tight_layout()
plt.show()

```

```python
Insights from the Monthly Sales Chart

Based on the example data:

📈 Overall sales are increasing from January to June. January has the lowest sales at 120. June has the highest sales at 220. Sales decreased slightly from February (150) to March (130). From March onward, sales show a consistent upward trend. The largest increase occurs from March to April and May to June. This suggests that sales performance is improving over time, with June being the strongest month.
```

4. Create a Bar Chart

Bar charts help compare values across categories.

Notes plt.bar() is used for categorical comparison. Each bar represents a category. Bar charts are easier to read when you compare totals or counts

```python
categories = ['A', 'B', 'C', 'D']
values = [25, 40, 30, 55]

plt.figure(figsize=(8, 5))
plt.bar(categories, values, color=['#4c72b0', '#55a868', '#c44e52', '#8172b2'])
plt.title('Product Sales by Category')
plt.xlabel('Category')
plt.ylabel('Sales')
for i, v in enumerate(values):
    plt.text(i, v + 2, str(v), ha='center')
plt.tight_layout()
plt.show()
```

Insights Category D has the highest sales: 55. Category B has the second-highest sales: 40. Category C has sales of 30. Category A has the lowest sales: 25. The difference between the highest and lowest categories is 30 units. Category D performs 2.2× better than Category A. Overall, Category D is the strongest-performing product category, while Category A may need attention or additional marketing

5. Create a Scatter Plot

A scatter plot shows the relationship between two numerical variables. Notes Scatter plots help reveal patterns, clusters, and outliers. If points rise together, there may be a positive relationship. This chart is useful for exploring correlation between variables.

```python
x = [10, 20, 30, 40, 50]
y = [20, 40, 60, 80, 100]
plt.figure(figsize=(8, 5))
plt.scatter(x, y, color='red', s=50, alpha=0.7)
plt.title('study hours vs Grades')
plt.xlabel('Hours studied')
plt.ylabel('Grades')
plt.grid(True)
plt.tight_layout()
plt.show()
```

Insights There is a positive relationship between study hours and scores. As study hours increase, scores generally increase. A student studying 1 hour scores 2, while studying 7 hours scores 15. The highest score is 15 at 7 hours of study. The points generally move upward, indicating that more study time is associated with better performance. There are no major downward movements in the data.



6. Create a Histogram

A histogram shows how data is distributed across value ranges. Notes A histogram groups values into bins. It helps show skewness, concentration, and spread. The number of bins affects how the pattern looks, so it should be chosen carefully.

```python
scores = [55, 60, 65, 66, 70, 72, 74, 75, 80, 82, 85, 90, 92, 95]
plt.figure(figsize=(8, 5))
plt.hist(scores, bins=5, color='blue', edgecolor='black')
plt.title('Test Scores')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.7)
plt.tight_layout()
plt.show()
```

Insights from the Exam Score Distribution

Distribution of Exam Scores

Frequency of students across score ranges. Scores range from 55 to 95.

The scores are fairly spread across the different ranges.

Most score ranges contain around 2–3 students, so there is no single dominant group.

The 55–62 range has the lowest frequency.

Higher scores, particularly 80 and above, are reasonably common.

The distribution does not show a strong concentration at either the very low or very high end.$0

Pie charts show how a total is divided into parts. Notes Pie charts are best for showing proportions, not detailed comparisons. They work best when there are only a few categories. Labels and percentages help explain the parts of the whole.

```python
labels = ['A','B','C','D']
sizes = [25,40,30,55]
plt.figure(figsize=(8,5))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, colors=['#4c72b0', '#55a868', '#c44e52', '#8172b2'])
plt.title('Product Sales by Category')
plt.tight_layout()
plt.show()
```

Insights from the Market Share Chart Market Share Distribution

Market share across four categories. Category A has the largest market share at 30%. Categories B and D each have 25%. Category C has the smallest share at 20%. A and C differ by 10 percentage points. B and D together account for 50% of the market. The market is fairly balanced, with no category having more than one-third of the total share.

8. Customize Plot Styling

Matplotlib allows a lot of styling to improve chart readability. Notes Colors, markers, and line styles make charts easier to read. Clean styling helps focus attention on the message of the data. Good design improves clarity without adding noise.

```python
plt.figure(figsize=(9, 5))
plt.plot([1, 2, 3, 4, 5], [10, 15, 13, 18, 22],
         color='red', linestyle='--', linewidth=2.5, marker='s', markersize=8)
plt.title('Styled Line Chart')
plt.xlabel('X Label')
plt.ylabel('Y Label')
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()


```

```python
Insights from the Styled Line Chart Styled Line Chart

Values across five ordered x-axis points. The overall trend is increasing, from 10 to 22. The value increases from 10 → 15 between X=1 and X=2. There is a small decline from 15 → 13 at X=3. After X=3, the values increase steadily: 13 → 18 → 22. The highest value is 22 at X=5. The lowest value is 10 at X=1. Overall increase = 12 units, or 120% from the starting value
```

9. Add Labels, Titles, and Legends

Good charts need clear labels and legends. Notes Titles and axis labels help explain what the chart is showing. Legends identify what each line or series represents. Clear labeling makes the chart understandable to others.

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
product_a = [20, 25, 23, 30, 35]
product_b = [15, 18, 22, 26, 31]

plt.figure(figsize=(8, 5))
plt.plot(months, product_a, marker='o', label='Product A')
plt.plot(months, product_b, marker='o', label='Product B', linestyle='--')
plt.title('Monthly Comparison')
plt.xlabel('Month')
plt.ylabel('Units Sold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

Insights from Monthly Comparison$0 Monthly Comparison

Comparison of units sold for Product A and Product B from January to May. Product A performs better than Product B in every month. Product A increases from 20 units in January to 35 units in May. Product B increases from 15 units to 31 units during the same period. Both products show an overall upward trend. Product A has a small decline in March: 25 → 23 units. Product B increases consistently every month. The gap between the products is largest in January (5 units) and May (4 units). May is the best-performing month for both products

All examples below use the tips dataset from Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

sns.set_theme(style='whitegrid')
tips = sns.load_dataset('tips')
print(tips.head())

# 1. Line plot: average bill by day
avg_bill = tips.groupby('day')['total_bill'].mean().sort_index()
plt.figure(figsize=(8, 4))
plt.plot(avg_bill.index, avg_bill.values, marker='o', color='royalblue', linewidth=2)
plt.title('1. Line Plot: Average Bill by Day')
plt.xlabel('Day')
plt.ylabel('Average Total Bill')
plt.grid(True, alpha=0.4)
plt.show()

# 2. Bar chart: average tip by day
avg_tip = tips.groupby('day')['tip'].mean().sort_index()
plt.figure(figsize=(8, 4))
plt.bar(avg_tip.index, avg_tip.values, color='green', edgecolor='black')
plt.title('2. Bar Chart: Average Tip by Day')
plt.xlabel('Day')
plt.ylabel('Average Tip')
plt.show()

# 3. Bar chart: average bill by sex
sex_bill = tips.groupby('sex')['total_bill'].mean()
plt.figure(figsize=(7, 4))
plt.bar(sex_bill.index, sex_bill.values, color=['pink', 'skyblue'])
plt.title('3. Bar Chart: Average Bill by Sex')
plt.xlabel('Sex')
plt.ylabel('Average Total Bill')
plt.show()

# 4. Histogram: total bill distribution
plt.figure(figsize=(8, 4))
plt.hist(tips['total_bill'], bins=15, color='purple', edgecolor='black')
plt.title('4. Histogram: Total Bill Distribution')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()

# 5. Scatter plot: total bill vs tip
plt.figure(figsize=(8, 4))
plt.scatter(tips['total_bill'], tips['tip'], alpha=0.7, color='darkorange')
plt.title('5. Scatter Plot: Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()

# 6. Scatter plot with colored sex groups
plt.figure(figsize=(8, 4))
for sex_group, group in tips.groupby('sex'):
    plt.scatter(group['total_bill'], group['tip'], label=sex_group, alpha=0.7)
plt.title('6. Scatter Plot: Bill vs Tip by Sex')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.legend()
plt.show()

# 7. Box plot: total bill by day
plt.figure(figsize=(8, 4))
plt.boxplot([tips[tips['day'] == d]['total_bill'] for d in ['Thur', 'Fri', 'Sat', 'Sun']],
            labels=['Thur', 'Fri', 'Sat', 'Sun'])
plt.title('7. Box Plot: Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()

# 8. Box plot: total bill by smoker status
plt.figure(figsize=(7, 4))
plt.boxplot([tips[tips['smoker'] == s]['total_bill'] for s in ['Yes', 'No']],
            labels=['Smoker', 'Non-Smoker'])
plt.title('8. Box Plot: Total Bill by Smoker Status')
plt.ylabel('Total Bill')
plt.show()

# 9. Violin plot: total bill by day
plt.figure(figsize=(8, 4))
plt.violinplot([tips[tips['day'] == d]['total_bill'] for d in ['Thur', 'Fri', 'Sat', 'Sun']],
               showmeans=True)
plt.xticks([1,2,3,4], ['Thur', 'Fri', 'Sat', 'Sun'])
plt.title('9. Violin Plot: Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()

# 10. Count plot: number of observations by day
counts = tips['day'].value_counts().reindex(['Sun', 'Sat', 'Thur', 'Fri'])
plt.figure(figsize=(8, 4))
plt.bar(counts.index, counts.values, color='coral')
plt.title('10. Count Plot: Number of Observations by Day')
plt.xlabel('Day')
plt.ylabel('Count')
plt.show()

# 11. Pie chart: smoker share
smoker_counts = tips['smoker'].value_counts()
plt.figure(figsize=(7, 7))
plt.pie(smoker_counts.values, labels=smoker_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('11. Pie Chart: Smoker Share')
plt.show()

# 12. Heatmap: correlation matrix
corr = tips[['total_bill', 'tip', 'size']].corr()
plt.figure(figsize=(6, 5))
plt.imshow(corr, cmap='coolwarm')
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns)
plt.yticks(range(len(corr.index)), corr.index)
plt.title('12. Heatmap: Correlation Matrix')
plt.show()


```

Overall Business Insights Higher bills generally lead to higher tips. Saturday and Sunday are important business days. Saturday has the largest number of customers in the dataset. Most bills are relatively small, with a few high-value transactions. Smoker status does not appear to be a major factor in bill size. Party size is related to total spending. Weekend customer activity appears particularly important for restaurant revenue.
