```python
#Create a list of 5 employee names and add a new employee using append().
a = ["viram","amit","rohit","ram","raju"]
a.append("lakshmi")
print(a)
```

```python
#Create a list of account numbers and add three new account numbers using extend().
b = [1,2,3,4,5]
b.extend([6,7,8])
print(b)
```

```python
#Insert a new customer at the beginning of a customer list using insert().
a = ["viram","amit","rohit","ram","raju"]
a.insert(3,"rakesh")
print(a)
```

```python
#Insert a transaction amount at index 3 in a list.
c = [ 100, 200, 300, 400, 500]
c.insert(2,600)
print(c)
```

```python
#Remove the customer "Rahul" from a customer list using remove().
a = ["viram","amit","rohit","ram","raju","rahul"]
a.remove("rahul")
print(a)
```

```python
#Count how many times the account type "Savings" appears in a list using count().
account = ["saving", "current", "saving"]
count = account.count("saving")
print(count)
```

```python
#Find the index of "Current" in a list of account types using index().
account = ["saving", "current", "saving"]
index = account.index("current")
print(index)
```

```python
#Reverse a list of transaction IDs using reverse().
transaction = [101, 102, 103, 104, 105]
transaction.reverse()
print(transaction)

```

```python
#Sort a list of account balances in ascending order using sort().
balance = [10, 23,54, 87]
balance.sort()
print(balance)
```

```python
#Sort a list of customer names in descending order using sort().
name = ["viram","amit","rohit","ram","raju"]
name.sort(reverse=True)
print(name)
```

```python
#Make a copy of a list using copy() and verify that modifying the copied list does not affect the original
a =[1,2,3,4,5,6]
b=a.copy()
b[2]=7
print(a,b)
```
