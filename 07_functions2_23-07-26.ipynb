
### Question 13:Using `filter()`, `map()`, and `reduce()`:

2. Increase every active employee's salary by 15%.
3. Calculate the total updated payroll




```python
employees = [
    {"id": 101, "name": "Rahul", "salary": 45000, "active": True},
    {"id": 102, "name": "Anil", "salary": 65000, "active": True},
    {"id": 103, "name": "Sneha", "salary": 72000, "active": False},
    {"id": 104, "name": "Kiran", "salary": 82000, "active": True},
    {"id": 105, "name": "Meena", "salary": 55000, "active": True},
    {"id": 106, "name": "Arjun", "salary": 98000, "active": False}
]
active_employees = list(filter(lambda x:x["active"]==True,employees))
increased_salary = list(map(lambda x:{"id":x["id"],"name":x["name"],"salary":x["salary"]*1.15},active_employees))
total_updated_payroll = reduce(lambda x,y:x+y["salary"],increased_salary,0)
print(active_employees)
print(increased_salary)
print(total_updated_payroll)
```

Question 15:Using `filter()`, `map()`, and `reduce()`:
1. Extract students scoring more than 70.
2. Extract only their marks.
3. Calculate the average marks



```python
students = [
    {"name": "Ravi", "marks": 92},
    {"name": "Kiran", "marks": 68},
    {"name": "Meena", "marks": 85},
    {"name": "Ajay", "marks": 48},
    {"name": "Sneha", "marks": 79},
    {"name": "Rahul", "marks": 95}
]
more_than_70 = list(filter(lambda x:x["marks"]>70,students))
marks = list(map(lambda x:x["marks"],more_than_70))
average = reduce(lambda x,y:x+y,marks)/len(marks)
print(more_than_70)
print(marks)
print(average)
```

Quuestion 16:Using `filter()`, `map()`, and `reduce()`:
1. Extract products currently in stock.
2. Calculate each product's inventory value using `price × stock`.
3. Calculate the total inventory value


```python
products = [
    {"name": "Laptop", "price": 55000, "stock": 5},
    {"name": "Mouse", "price": 600, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 0},
    {"name": "Monitor", "price": 18000, "stock": 8},
    {"name": "Printer", "price": 12000, "stock": 0},
    {"name": "SSD", "price": 6500, "stock": 12}
]
in_stock = list(filter(lambda x:x["stock"]>0,products))
inventory_value = list(map(lambda x:{"name":x["name"],"inventory_value":x["price"]*x["stock"]},in_stock))
total_inventory_value = reduce(lambda x,y:x+y["inventory_value"],inventory_value,0)
print(in_stock)
print(inventory_value)
print(total_inventory_value)
```

Question 19:Using `filter()`, `map()`, and `reduce()`:
1. Extract transactions whose amount is greater than 5000.
2. Add 18% GST to every selected transaction.
3. Calculate the grand total after GST.





```python
transactions = [
    {"id": 1, "amount": 1200},
    {"id": 2, "amount": 5200},
    {"id": 3, "amount": 18000},
    {"id": 4, "amount": 2500},
    {"id": 5, "amount": 9600},
    {"id": 6, "amount": 7200}
]
greater_than_5000 = list(filter(lambda x:x["amount"]>5000,transactions))
gst = list(map(lambda x:{"id":x["id"],"amount":x["amount"]*1.18},greater_than_5000))
grand_total = reduce(lambda x,y:x+y,map(lambda x:x["amount"],gst))
print(greater_than_5000)
print(gst)
print(grand_total)


```

Question 20:Using `filter()`, `map()`, and `reduce()`:
1. Extract active salespeople from the South region.
2. Calculate a 5% incentive for each selected salesperson.
3. Create a list containing salesperson name, sales amount, and incentive.
4. Calculate the total incentive payable.
 total incentive payable.

```python
sales_records = [
    {"salesperson": "Rahul", "region": "South", "sales": 45000, "active": True},
    {"salesperson": "Anil", "region": "North", "sales": 65000, "active": True},
    {"salesperson": "Sneha", "region": "South", "sales": 72000, "active": False},
    {"salesperson": "Kiran", "region": "South", "sales": 82000, "active": True},
    {"salesperson": "Meena", "region": "West", "sales": 55000, "active": True},
    {"salesperson": "Arjun", "region": "South", "sales": 98000, "active": True}
]
south_region = list(filter(lambda x: x["region"]=="South" and x["active"]==True, sales_records))
incentive = list(map(lambda x: {"salesperson": x["salesperson"], "sales": x["sales"], "incentive": x["sales"]*0.05}, south_region))
salesperson_sales_incentive = list(map(lambda x: {"salesperson": x["salesperson"], "sales": x["sales"], "incentive": x["incentive"]}, incentive))
total_incentive = reduce(lambda x, y: x+y["incentive"], salesperson_sales_incentive, 0)
print(south_region)
print(incentive)
print(salesperson_sales_incentive)
print(total_incentive)


```
