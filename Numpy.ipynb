You are provided with two main NumPy arrays:

```python
import numpy as np
stores = np.array([
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Mumbai",
    "Pune"
])
categories = np.array([
    "Electronics",
    "Groceries",
    "Clothing",
    "Furniture",
    "Beauty",
    "Sports"
])
sales_data = np.array([
    [850000, 420000, 310000, 275000, 190000, 225000],
    [780000, 460000, 295000, 310000, 210000, 240000],
    [920000, 390000, 340000, 260000, 185000, 215000],
    [1050000, 510000, 375000, 345000, 250000, 290000],
    [690000, 405000, 280000, 225000, 175000, 205000]
])
cost_data = np.array([
    [640000, 315000, 220000, 205000, 125000, 160000],
    [590000, 350000, 210000, 225000, 140000, 170000],
    [700000, 295000, 245000, 195000, 120000, 155000],
    [790000, 390000, 265000, 250000, 165000, 205000],
    [530000, 310000, 205000, 170000, 115000, 145000]
])
print("Stores:")
print(stores)
print("\nProduct Categories:")
print(categories)
print("\nSales Data:")
print(sales_data)
print("\nCost Data:")
print(cost_data)
```

```python
import numpy as np
stores = np.array([
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Mumbai",
    "Pune"
])
categories = np.array([
    "Electronics",
    "Groceries",
    "Clothing",
    "Furniture",
    "Beauty",
    "Sports"
])
sales_data = np.array([
    [850000, 420000, 310000, 275000, 190000, 225000],
    [780000, 460000, 295000, 310000, 210000, 240000],
    [920000, 390000, 340000, 260000, 185000, 215000],
    [1050000, 510000, 375000, 345000, 250000, 290000],
    [690000, 405000, 280000, 225000, 175000, 205000]
])
cost_data = np.array([
    [640000, 315000, 220000, 205000, 125000, 160000],
    [590000, 350000, 210000, 225000, 140000, 170000],
    [700000, 295000, 245000, 195000, 120000, 155000],
    [790000, 390000, 265000, 250000, 165000, 205000],
    [530000, 310000, 205000, 170000, 115000, 145000]
])
total_sales = np.sum(sales_data, axis=1)
total_cost = np.sum(cost_data, axis=1)
profit = total_sales - total_cost
overall_profit_margin = np.mean(profit / total_sales)
average_transaction_value = np.mean(total_sales)
highest_individual_sales_value = np.max(total_sales)
lowest_individual_sales_value = np.min(total_sales)
total_sales_by_store = np.sum(sales_data, axis=0)
total_cost_by_store = np.sum(cost_data, axis=0)
total_profit_by_store = total_sales_by_store - total_cost_by_store
profit_margin_by_store = total_profit_by_store / total_sales_by_store
number_of_category_targets_achieved = np.sum(sales_data >= 400000, axis=0)
total_units_sold_by_store = np.sum(sales_data, axis=1)
expansion_eligibility_by_store = total_units_sold_by_store >= 5
total_sales_by_category = np.sum(sales_data, axis=1)
total_cost_by_category = np.sum(cost_data, axis=1)
total_profit_by_category = total_sales_by_category - total_cost_by_category
profit_margin_by_category = total_profit_by_category / total_sales_by_category
average_sales_by_category = np.mean(sales_data, axis=0)
number_of_stores_achieving_target = np.sum(sales_data >= 400000, axis=1)
total_units_sold_by_category = np.sum(sales_data, axis=0)
best_performing_store_index = np.argmax(total_sales_by_store)
lowest_performing_store_index = np.argmin(total_sales_by_store)
most_profitable_category_index = np.argmax(total_profit_by_category)
lowest_profitable_category_index = np.argmin(total_profit_by_category)
store_recommended_for_expansion_index = np.argmax(expansion_eligibility_by_store)
category_requiring_additional_marketing_index = np.argmax(number_of_stores_achieving_target)
store_category_combinations_requiring_cost_optimization_indices = np.where(sales_data < 400000)
store_category_combinations_requiring_inventory_reorder_indices = np.where(sales_data < 300000)
print("Company Summary")
print("Total sales: ", total_sales)
print("Total cost: ", total_cost)
print("Total profit: ", profit)
print("Overall profit margin: ", overall_profit_margin)
print("Average transaction value: ", average_transaction_value)
print("Highest individual sales value: ", highest_individual_sales_value)
print("Lowest individual sales value: ", lowest_individual_sales_value)
print("Store Summary")
print("Total sales of Store: ", total_sales_by_store)
print("Total cost of Store: ", total_cost_by_store)
print("Total profit of Store: ", total_profit_by_store)
print("Profit margin of Store: ", profit_margin_by_store)
print("Number of category targets achieved: ", number_of_category_targets_achieved)
print("Total units sold by Store: ", total_units_sold_by_store)
print("Expansion eligibility of Store: ", expansion_eligibility_by_store)
print("Category saummary")
print("Total sales of Category: ", total_sales_by_category)
print("Total cost of Category: ", total_cost_by_category)
print("Total profit of Category: ", total_profit_by_category)
print("Profit margin of Category: ", profit_margin_by_category)
print("Average sales of Category: ", average_sales_by_category)
print("Number of stores achieving target: ", number_of_stores_achieving_target)
print("Total units sold by Category: ", total_units_sold_by_category)
print("Management Recommendations")
print("Best performing store: ", stores[best_performing_store_index])
print("Lowest performing store: ", stores[lowest_performing_store_index])
print("Most profitable category: ", categories[most_profitable_category_index])
print("Lowest profitable category: ", categories[lowest_profitable_category_index])
print("Store recommended for expansion: ", stores[store_recommended_for_expansion_index])
print("Category requiring additional marketing: ", categories[category_requiring_additional_marketing_index])
print("Store-category combinations requiring cost optimization: ", store_category_combinations_requiring_cost_optimization_indices)
print("Store-category combinations requiring inventory reorder: ", store_category_combinations_requiring_inventory_reorder_indices)
print("SMARTMART RETAIL PERFORMANCE REPORT")
print("==================================")
print("COMPANY SUMMARY")
print("---------------")
print("Total Sales          : ", total_sales)
print("Total Cost           : ", total_cost)
print("Total Profit         : ", profit)
print("Overall Profit Margin: ", overall_profit_margin)
print()
print("BEST-PERFORMING STORE")
print("--------------------")
print("Store        : ", stores[best_performing_store_index])
print("Sales        : ", total_sales_by_store[best_performing_store_index])
print("Cost         : ", total_cost_by_store[best_performing_store_index])
print("Profit       : ", total_profit_by_store[best_performing_store_index])
print("Profit Margin: ", profit_margin_by_store[best_performing_store_index])
print()
print("MOST PROFITABLE CATEGORY")
print("-----------------------")
print("Category     : ", categories[most_profitable_category_index])
print("Sales        : ", total_sales_by_category[most_profitable_category_index])
print("Cost         : ", total_cost_by_category[most_profitable_category_index])
print("Profit       : ", total_profit_by_category[most_profitable_category_index])
print("Profit Margin: ", profit_margin_by_category[most_profitable_category_index])
print("EXPANSION RECOMMENDATIONS")
print("------------------------")
print("Eligible Stores: ", stores[expansion_eligibility_by_store])
print()
print("COST OPTIMIZATION")
print("----------------")
print("Store        : ", stores[store_category_combinations_requiring_cost_optimization_indices[0]])
print("Category     : ", categories[store_category_combinations_requiring_cost_optimization_indices[1]])
print("Cost         : ", cost_data[store_category_combinations_requiring_cost_optimization_indices[0], store_category_combinations_requiring_cost_optimization_indices[1]])
print("Profit Margin: ", profit_margin_by_category[store_category_combinations_requiring_cost_optimization_indices[0]])
print()
print("INVENTORY REORDER")
print("----------------")
print("Store           : ", stores[store_category_combinations_requiring_inventory_reorder_indices[0]])
print("Category        : ", categories[store_category_combinations_requiring_inventory_reorder_indices[1]])
print("Closing Stock   : ", sales_data[store_category_combinations_requiring_inventory_reorder_indices[0], store_category_combinations_requiring_inventory_reorder_indices[1]])
print("Reorder Quantity: ", 100 - sales_data[store_category_combinations_requiring_inventory_reorder_indices[0], store_category_combinations_requiring_inventory_reorder_indices[1]])
```
