### Q8. Protect a amount with a property for banking case 1; use sample reference `ACC-1008` and explain the result.

```python
class Account:
    def __init__(self, account_number, amount):
        self.account_number = account_number
        self.amount = amount
    @property
    def amount(self):
        return self._amount
    @amount.setter
    def amount(self, value):
        if value < 0:
            raise ValueError("Amount cannot be negative.")
        self._amount = value
account = Account("ACC-1008", 5000)
print("Account:", account.account_number)
print("Amount:", account.amount)
account.amount = 6000
print("Updated amount:", account.amount)

```

### Q11. Use a static method to validate account reference for banking case 1; use sample reference `ACC-1011` and explain the result.

```python
class Account:
    def __init__(self, account_number):
        self.account_number = account_number
    @staticmethod
    def validate_account_reference(reference):
        return (
            isinstance(reference, str)
            and reference.startswith("ACC-")
            and reference[4:].isdigit()
        )
account = Account("ACC-1011")
print("Account:", account.account_number)
print("Valid reference:",
      Account.validate_account_reference(account.account_number))

```
