### Q31. Parse a receipt line into named groups for banking case 1; use sample reference `ACC-1031` and explain the result.

**Hints**

1. Write representative valid and invalid samples before writing the pattern.
2. Use raw strings, named groups for extraction, and `fullmatch` for whole-field validation.
3. Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
import re
text = 'ACC-1031 deposit=₹2500'
pattern = re.compile(
    r'(?P<account>ACC-\d{4})\s+deposit=₹(?P<amount>\d+(?:\.\d+)?)'
)
match = pattern.fullmatch(text)
if match:
    print('Account:', match.group('account'))
    print('Amount:', match.group('amount'))
else:
    print('Invalid receipt line')

```

### Q100. Create a regex-based data quality report for banking case 3; use sample reference `ACC-1100` and explain the result.

**Hints**

1. Write representative valid and invalid samples before writing the pattern.
2. Use raw strings, named groups for extraction, and `fullmatch` for whole-field validation.
3. Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
import re
account_pattern = re.compile(r'^ACC-\d{4}$')
amount_pattern = re.compile(r'^(?:INR|USD|EUR)\s+\d+(?:\.\d{2})?$')
transfer_pattern = re.compile(r'^TXN-\d{5}$')
accounts = [
    "ACC-1234",
    "ACC-5678",
    "BAD-1234",
    ""
]
amounts = [
    "INR 1000.00",
    "USD 250.50",
    "GBP 100.00",
    "EUR abc"
]
transfers = [
    "TXN-12345",
    "TXN-67890",
    "BAD-12345"
]
def quality_report(values, pattern):
    valid = sum(bool(pattern.fullmatch(value)) for value in values)
    return {
                "total": len(values),
        "valid": valid,
        "invalid": len(values) - valid
    }
print("Regex Data Quality Report")
print("-------------------------")
print("Accounts :", quality_report(accounts, account_pattern))
print("Amounts  :", quality_report(amounts, amount_pattern))
print("Transfers:", quality_report(transfers, transfer_pattern))


```
